```sql
CREATE OR REFRESH MATERIALIZED VIEW  gold_customer_summary
AS
SELECT
c.customer_id,
c.customer_name,
c.date_of_birth,
c.telephone,
c.email,
a.address_line_1,
a.city,
a.state,
a.postcode,
COUNT(DISTINCT o.order_id) AS total_orders,
SUM(o.quantity)as total_items_ordered,
SUM(o.price * o.quantity) as total_order_amount
FROM LIVE.silver_customers c
INNER JOIN LIVE.silver_order o
ON c.customer_id = o.customer_id
INNER JOIN LIVE.silver_address a
ON c.customer_id = a.customer_id 
WHERE a.__END_AT IS NULL
GROUP BY ALL;
```
