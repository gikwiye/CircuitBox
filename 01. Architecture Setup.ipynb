#### Create CircuitBox Catalog

```sql
CREATE EXTERNAL LOCATION landing_location
URL 'abfss://circuitbox@databrickscoursesng.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL databricks_course_ext_sc);
```

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
MANAGED LOCATION 'abfss://circuitbox@databrickscoursesng.dfs.core.windows.net';
```

#### Create Landing and lakehouse schema

```sql
CREATE SCHEMA IF NOT EXISTS circuitbox.landing
MANAGED LOCATION 'abfss://circuitbox@databrickscoursesng.dfs.core.windows.net/landing';
```

```sql
CREATE SCHEMA IF NOT EXISTS circuitbox.lakehouse
MANAGED LOCATION 'abfss://circuitbox@databrickscoursesng.dfs.core.windows.net/lakehouse';
```

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS circuitbox.landing.operational_data
LOCATION 'abfss://circuitbox@databrickscoursesng.dfs.core.windows.net/landing/operational_data'
COMMENT 'Operational data for circuitbox';
```
